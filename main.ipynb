```python
df = pd.read_excel('bd.xlsx')
df
```

```python
def ajouter_these(df, nom, prenom, impertant, contact, theme, thematique, domaine, filiere, entite, directeur, date_depot, date_soutenance, mentions):
    """
    Ajoute une nouvelle thèse au DataFrame.
    
    Args:
        df (pd.DataFrame): DataFrame à modifier
        impertant (str): Nom de l'impétrant
        contact (str): Contact (email/téléphone)
        theme (str): Thème de la thèse
        thematique (str): Thématique
        domaine (str): Domaine de recherche
        filiere (str): Filière
        entite (str): Entité de rattachement
        directeur (str): Directeur de thèse
        date_depot (str): Date de dépôt (format JJ/MM/AAAA ou AAAA-MM-JJ)
        date_soutenance (str): Date de soutenance (format JJ/MM/AAAA ou AAAA-MM-JJ)
        mentions (str): Mentions spéciales
    
    Returns:
        pd.DataFrame: DataFrame mis à jour
    """
    nouvelle_ligne = {
        "Date de dépôt": pd.to_datetime(date_depot, dayfirst=True),
        "Nom": nom,
        "Prenom": prenom,
        "Impétrant": impertant,
        "Contact": contact,
        "Theme": theme,
        "Thématique": thematique,
        "Domaine": domaine,
        "Filière": filiere,
        "Entité": entite,
        "Directeur": directeur,
        "Date de soutenance": pd.to_datetime(date_soutenance, dayfirst=True),
        "Mentions": mentions
    }
    
    return pd.concat([df, pd.DataFrame([nouvelle_ligne])], ignore_index=True)
```

```python
df = ajouter_these(
    df=df,
    nom="BATIONO",
    prenom="Marie Prudence Karidja",
    impertant="BATIONO Marie Prudence Karidja",
    contact="",
    theme="La gestion du risque opérationnel de la caisse: cas de AEB-CI",
    thematique="Comptabilité",
    domaine="Comptabilité",
    filiere="Audit et contrôle",
    entite="caritas CI",
    directeur="",
    date_depot="01/12/20",  # Format JJ/MM/AAAA
    date_soutenance="2020-12-21",  # Format AAAA-MM-JJ
    mentions="Bien"
)
df = ajouter_these(
    df=df,
    nom="N'GUESSAN",
    prenom="Amenan Laura",
    impertant="N'GUESSAN Amenan Laura",
    contact="",
    theme="Audit des comptes fournisseurs d'une société immobilère: cas de la société ALPHA",
    thematique="Comptabilité",
    domaine="Comptabilité",
    filiere="Audit et contrôle",
    entite="caritas CI",
    directeur="",
    date_depot="02/12/20",  # Format JJ/MM/AAAA
    date_soutenance="2020-12-22",  # Format AAAA-MM-JJ
    mentions="Très Bien"
)
df
```
