```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
economy_df = pd.read_excel("Economie_du_developpement.xlsx")
economy_df.head(2)
```

```python
economy_df.groupby('Domaine').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(False)
```

```python
Domaine = economy_df['Domaine'].unique()
Domaine
```

```python
Domaine = economy_df['Domaine'].value_counts()
Domaine
```

```python
economy_df.groupby('Domaine').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(False)

plt.title('Répartition par Domaine en économie du développement', pad=20, fontsize=14, fontweight='bold')
plt.xlabel('Nombre de travaux', labelpad=10)
plt.ylabel('domaine', labelpad=10)
```

```python
# Configuration du style
#plt.style.use('seaborn')  # ou 'ggplot' pour un autre style esthétique
plt.figure(figsize=(10, 6))  # Taille du graphique

# Création du diagramme
economy_df.groupby('Thématique').size().sort_values().plot(
    kind='barh',
    color=sns.color_palette('Dark2'),
    edgecolor='black',
    linewidth=0.5
)

# Personnalisation
plt.title('Répartition par Thématique en économie du développement', pad=20, fontsize=14, fontweight='bold')
plt.xlabel('Nombre de travaux', labelpad=10)
plt.ylabel('Thématique', labelpad=10)

# Suppression des bordures inutiles
for spine in ['top', 'right', 'left']:
    plt.gca().spines[spine].set_visible(False)

# Ajout d'une grille discrète
plt.grid(axis='x', linestyle='--', alpha=0.4)

# Ajustement des marges
plt.tight_layout()

# Affichage
plt.show()
```

```python
# Calcul des données (identique à votre code)
domain_counts = economy_df.groupby('Domaine').size()

# Création du diagramme circulaire
plt.figure(figsize=(10, 8))
domain_counts.plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=sns.palettes.mpl_palette('Dark2'),  # Même palette de couleurs
    wedgeprops={'linewidth': 1, 'edgecolor': 'white'},  # Contours blancs
    startangle=90
)

# Personnalisation du style
plt.title('Répartition par domaine', pad=20, fontweight='bold')
plt.ylabel('')  # Supprime l'étiquette 'Domaine' par défaut
plt.gca().axis('equal')  # Rend le cercle parfait

# Suppression des axes (comme dans votre code barh)
plt.gca().spines[['top', 'right', 'bottom', 'left']].set_visible(False)

plt.tight_layout()
plt.show()
```
