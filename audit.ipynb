```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
audit_df = pd.read_excel('audit_et_controle.xlsx')
audit_df.head(2)
```

```python
Domaine = audit_df['Domaine'].value_counts()
Domaine
```

```python
audit_df.groupby('Domaine').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(False)
```

```python
# Configuration du style
#plt.style.use('seaborn')  # ou 'ggplot' pour un autre style esthétique
plt.figure(figsize=(10, 6))  # Taille du graphique

# Création du diagramme
audit_df.groupby('Domaine').size().sort_values().plot(
    kind='barh',
    color=sns.color_palette('Dark2'),
    edgecolor='black',
    linewidth=0.5
)

# Personnalisation
plt.title('Répartition par domaine', pad=20, fontsize=14, fontweight='bold')
plt.xlabel('Nombre', labelpad=10)
plt.ylabel('Domaine', labelpad=10)

# Suppression des bordures inutiles
for spine in ['top', 'right', 'left']:
    plt.gca().spines[spine].set_visible(False)

# Ajout d'une grille discrète
plt.grid(axis='x', linestyle='--', alpha=0.4)

# Ajustement des marges
plt.tight_layout()

# Affichage
plt.show()
```

```python
# Calcul des données (identique à votre code)
domain_counts = audit_df.groupby('Domaine').size()

# Création du diagramme circulaire
plt.figure(figsize=(10, 8))
domain_counts.plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=sns.palettes.mpl_palette('Dark2'),  # Même palette de couleurs
    wedgeprops={'linewidth': 1, 'edgecolor': 'white'},  # Contours blancs
    startangle=90
)

# Personnalisation du style
plt.title('Répartition par domaine', pad=20, fontweight='bold')
plt.ylabel('')  # Supprime l'étiquette 'Domaine' par défaut
plt.gca().axis('equal')  # Rend le cercle parfait

# Suppression des axes (comme dans votre code barh)
plt.gca().spines[['top', 'right', 'bottom', 'left']].set_visible(False)

plt.tight_layout()
plt.show()
```
