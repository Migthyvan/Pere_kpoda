```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
bank_df = pd.read_excel("Banque_et_assurance.xlsx")
bank_df.head(2)
```

```python
bank_df.groupby('Domaine').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(False)
```

```python
bank_df['Domaine'] = bank_df['Domaine'].replace({
    'Gestion du risque bancaire': 'Gestion des risques bancaires',
})
```

```python
Domaine = bank_df['Domaine'].unique()
Domaine
```

```python
bank_df.groupby('Domaine').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(False)

plt.title('Répartition par domaine en banque et assurance', pad=20, fontsize=14, fontweight='bold')
plt.xlabel('Nombre de travaux', labelpad=10)
plt.ylabel('domaine', labelpad=10)
```

```python
bank_df.to_excel('Banque_et_assurance.xlsx', index = False)
```

```python
# Configuration du style
#plt.style.use('seaborn')  # ou 'ggplot' pour un autre style esthétique
plt.figure(figsize=(10, 6))  # Taille du graphique

# Création du diagramme
bank_df.groupby('Thématique').size().sort_values().plot(
    kind='barh',
    color=sns.color_palette('Dark2'),
    edgecolor='black',
    linewidth=0.5
)

# Personnalisation
plt.title('Répartition par Thématique en banque et assurance', pad=20, fontsize=14, fontweight='bold')
plt.xlabel('Nombre de travaux', labelpad=10)
plt.ylabel('Thématique', labelpad=10)

# Suppression des bordures inutiles
for spine in ['top', 'right', 'left']:
    plt.gca().spines[spine].set_visible(False)

# Ajout d'une grille discrète
plt.grid(axis='x', linestyle='--', alpha=0.4)

# Ajustement des marges
plt.tight_layout()

# Affichage
plt.show()
```
